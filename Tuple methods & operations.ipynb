### Tuple methods: 
- tuples in python have only two built-in methods(since tuples are immutable).
- python keeps them lightweight.

### count(): returns the number of times a specific value appears in the tuple. 

```python
this_tuple = (1, 3, 7, 8, 7, 5, 4, 6, 8, 5)
x = this_tuple.count(5)
print(x)
```

### index(): finds the first occurrence of specific value, return its position. Raises an exception if specific value is not found.

```python
this_tuple = (1, 3, 7, 8, 7, 5, 4, 6, 8, 5)
x = this_tuple.index(8)
print(x)
```

### Other useful operations(not methods, but built-in)
- since tuples are immutable, they don't have append(), remove(), etc you can use:

```python
#concatenation
x = (1,2) + (3,4)
print(x)
```

```python
#repetition
x = (1,2) * 2
print(x)
```

```python
#membership
x = 2 in (1,2,3)
print(x)
```

```python
#length
len((1,2,3))
```

```python
#max/min
max((1,2,3))
```

```python
#sum
sum((1,2,3))
```

```python
#slicing
x = (1,2,3,4)
x[1:3]
```

```python
#unpacking
a, b = (1, 2)
print(a)
print(b)
```

```python
#conversion
x = tuple([1,2,3])
print(type(x))
```
